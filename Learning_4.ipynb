# Git Commands in WSL (Windows Subsystem for Linux)

## 1. Navigate to the D: drive

```bash
cd /mnt/d
```

**Explanation:**
- `cd` means change directory.
- `/mnt/d` is the D: drive mounted in WSL.
- This command moves you to the D: drive so you can work with your files.

## 2. Set the default branch name

```bash
git config --global init.defaultBranch main
```

**Explanation:**
- `git config` changes Git settings.
- `--global` applies the setting to all repositories for your WSL user.
- `init.defaultBranch main` makes `main` the default branch name for new repositories.

## 3. Initialize a Git repository

```bash
git init
```

**Explanation:**
- Creates a new Git repository in the current directory.
- Git creates a hidden `.git` folder to store version history and configuration.
- Run this inside the project folder you want to track.

## 4. Add files to the staging area

```bash
git add newfile
```

Or add all new and modified files in the current directory and its subdirectories:

```bash
git add .
```

**Explanation:**
- `git add newfile` stages a specific file named `newfile`.
- `git add .` stages changes for the next commit.
- The staging area lets you choose which changes to save in a commit.

**Note:** Replace `newfile` with the actual filename. The file must exist.

## 5. Configure your Git email

```bash
git config --global user.email "you@example.com"
```

**Explanation:**
- Sets the email address associated with your commits.
- Replace `you@example.com` with your email address.

## 6. Configure your Git username

```bash
git config --global user.name "Your Name"
```

**Explanation:**
- Sets the author name recorded in your commits.
- Replace `Your Name` with your preferred name.

## 7. Commit staged changes

```bash
git commit -m "added newfile"
```

**Explanation:**
- Saves the staged changes as a commit in the repository's history.
- `-m` specifies the commit message.
- `"added newfile"` describes what you changed.

**Important:** Run `git add` before `git commit`. Git also requires at least one staged change to make a normal commit.

## 8. Create a new branch

```bash
git branch my1stbranch
```

**Explanation:**
- Creates a branch named `my1stbranch`.
- A branch allows you to work on a separate version of your project.
- This command creates the branch but does not switch to it.

## 9. Create and switch to a new branch

```bash
git checkout -b my1stbranch
```

**Explanation:**
- `git checkout` switches branches.
- `-b` creates a new branch and switches to it immediately.

**Note:** If you already created `my1stbranch` with `git branch`, use `git checkout my1stbranch` instead. Otherwise, Git reports that the branch already exists.

## 10. Display all local branches

```bash
git branch
```

**Explanation:**
- Lists the local branches in your repository.
- An asterisk (`*`) identifies the current branch.

## 11. Switch to another branch

```bash
git checkout my1stbranch
```

**Explanation:**
- Switches your working directory to `my1stbranch`.
- You can now work on that branch separately from `main`.

## 12. Switch back to the main branch

```bash
git checkout main
```

**Explanation:**
- Switches the current branch to `main`.
- Use this when you want to integrate changes from another branch into the main branch.

## 13. Merge a branch into the current branch

```bash
git merge my1stbranch
```

**Explanation:**
- Merges the changes from `my1stbranch` into the branch you currently have checked out.
- To merge into `main`, switch to `main` first.
- Git may perform a fast-forward merge or create a merge commit.
- If conflicts occur, resolve them before completing the merge.

## Important WSL Notes

- In WSL, run Git commands directly in the Linux terminal; you do not need `!` before commands.
- Use `pwd` to display your current directory.
- Use `ls` to list files and folders.
- Use `git status` to check the repository's current state.
- Use `cd /mnt/c` for the Windows C: drive and `cd /mnt/d` for the Windows D: drive, assuming those drives are mounted.
- Run `git init` inside your actual project directory. Initializing in `/mnt/d` itself may accidentally make the entire drive a repository.

## Recommended Git Workflow

1. Navigate to your project directory.
2. Run `git init`.
3. Configure your Git username and email.
4. Create or edit files.
5. Run `git add .`.
6. Run `git commit -m "Initial commit"`.
7. Create a branch with `git checkout -b my1stbranch`.
8. Make changes and commit them on the new branch.
9. Switch to `main` with `git checkout main`.
10. Merge the branch with `git merge my1stbranch`.

**Remember:** `git add` stages changes, `git commit` saves them in history, `git branch` manages branches, `git checkout` switches branches, and `git merge` combines branch changes.

# Using GitHub with Git in WSL

## 1. Go to the project folder

```bash
cd /mnt/d
```

**Explanation:** Moves to the Windows D: drive in WSL. You can replace `/mnt/d` with your actual project directory.

## 2. Save the GitHub repository URL in a variable

```bash
export ORIGIN=https://github.com/miladghavami/gkpbt-css-circle.git
```

**Explanation:**
- `export` creates an environment variable called `ORIGIN`.
- The variable stores the URL of your GitHub repository.
- `$ORIGIN` lets you use the saved URL in later commands.
- This variable remains available in the current shell session.

## 3. Clone your GitHub repository

```bash
git clone "$ORIGIN"
```

**Explanation:**
- `git clone` downloads a copy of a remote repository to your computer.
- `"$ORIGIN"` supplies the repository URL stored in the variable.
- Git creates a new folder for the cloned repository.

Enter the cloned folder afterward:

```bash
cd gkpbt-css-circle
```

## 4. See the changes you made to a file

```bash
git diff ./style.css
```

**Explanation:**
- `git diff` displays changes that have not yet been staged.
- `./style.css` refers to the `style.css` file in the current directory.
- Lines beginning with `-` are removed lines; lines beginning with `+` are added lines.
- If the file has no unstaged changes, the command may display nothing.

To see staged changes instead, use:

```bash
git diff --cached
```

## 5. Configure your Git email

```bash
git config --global user.email "your-email@example.com"
```

**Explanation:**
- Sets the email address Git records in your commits.
- `--global` applies the setting to your WSL user across repositories.
- Replace the example email with your own Git commit email.

## 6. Configure your Git username

```bash
git config --global user.name "Your Name"
```

**Explanation:**
- Sets the author name recorded in your commits.
- Replace `"Your Name"` with your preferred name.

## 7. Push a branch to GitHub

```bash
git push -u origin bug-fix-typo
```

**Explanation:**
- `git push` uploads local commits to GitHub.
- `origin` is the default remote name often created by `git clone`.
- `bug-fix-typo` is the local branch you want to push.
- `-u` sets the upstream branch, connecting your local branch to its remote counterpart for future pushes and pulls.

**Important:** The branch must exist locally and contain commits to push. If your branch has another name, replace `bug-fix-typo` with that name.

## 8. Open the current folder in Visual Studio Code

```bash
code .
```

**Explanation:**
- `code` launches Visual Studio Code from the terminal.
- `.` means the current directory.
- This works when VS Code and its WSL integration are installed and configured.

## 9. Delete a folder

```bash
rm -r my_folder
```

**Explanation:**
- `rm` removes files or directories.
- `-r` means recursive, so it removes a directory and its contents.
- `my_folder` is the directory to delete.

**Warning:** This deletion is generally permanent. Check the folder name before running the command. If you see `No such file or directory`, the folder does not exist at the current path or has a different name.

## 10. View a compact Git history

```bash
git log --oneline
```

**Explanation:**
- Displays the repository's commit history.
- `--oneline` shows each commit on one compact line.
- Each line includes a shortened commit hash and its commit message.

Example:

```text
e13953e Fix typo in CSS
a12bc45 Add circle styling
```

The hashes above are examples only.

## 11. Revert a specific commit

```bash
git revert e13953e
```

**Explanation:**
- Creates a new commit that reverses the changes introduced by the specified commit.
- `e13953e` is an example of a shortened commit hash; replace it with the real hash from `git log --oneline`.
- Unlike resetting history, `git revert` preserves the existing commit history.

**Important:** Git may open an editor for the revert commit message. If there are conflicts, resolve them before completing the revert.

## Important: WSL versus Google Colab

These commands are intended for the WSL terminal. In WSL, run commands directly without `!`.

For example:

```bash
git status
git diff ./style.css
git log --oneline
```

If you enter `git clone "$ORIGIN"` directly into a normal Python code cell in Google Colab, Python reports `SyntaxError` because it interprets the command as Python code. In Colab, use `!git clone "$ORIGIN"` for a shell command or use `%cd` to change directories.

## Recommended GitHub workflow

1. Clone the repository with `git clone`.
2. Enter the project folder with `cd`.
3. Edit your files.
4. Inspect changes with `git status` and `git diff`.
5. Stage changes with `git add .`.
6. Commit changes with `git commit -m "Describe your changes"`.
7. Push your branch with `git push -u origin your-branch-name`.
8. Check commit history with `git log --oneline`.

**Remember:** `clone` downloads a repository, `diff` displays changes, `push` uploads commits, `log` displays history, and `revert` creates a commit that reverses an earlier commit.
