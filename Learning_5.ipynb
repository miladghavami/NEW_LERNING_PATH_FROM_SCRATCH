# 📊 Descriptive Statistics: Graphs, Center, and Spread

## 1. Types of Data and Graphs

### Categorical Data
Categorical data represents groups or categories.

- **Pie Chart:** Shows the proportion or percentage of each category.
- **Bar Chart:** Compares the frequency or count of different categories.

### Quantitative Data
Quantitative data consists of numerical values.

- **Histogram:** Shows the distribution of numerical data by grouping values into intervals (bins).
- **Boxplot:** Displays the center, spread, and potential outliers using the five-number summary.

---

## 2. How to Describe a Histogram

When describing a histogram, consider the following characteristics:

### A. Shape
- **Symmetric:** Both sides have approximately the same shape.
- **Right-skewed (positively skewed):** The distribution has a longer tail on the right.
- **Left-skewed (negatively skewed):** The distribution has a longer tail on the left.
- **Bell-shaped:** A roughly symmetric, mound-shaped distribution.

### B. Modality (Number of Peaks)
- **Unimodal:** One peak.
- **Bimodal:** Two peaks.
- **Multimodal:** More than two peaks.

### C. Center
The center describes the typical or central value of the data.

- **Mean:** The arithmetic average of all observations.
- **Median:** The middle value when observations are arranged from smallest to largest.

### D. Spread
Spread describes how much the data values vary.

- **Minimum:** The smallest value.
- **Maximum:** The largest value.
- **Range:** The difference between the maximum and minimum.

  Formula: Range = Maximum − Minimum

- **Standard Deviation (SD):** Measures how spread out the observations are around the mean.
  - Low SD: Values tend to be close to the mean.
  - High SD: Values tend to be farther from the mean.

### E. Outliers
Outliers are observations that are unusually far from the rest of the data.

### F. Most Common Values
Identify where the histogram has the highest bars. These intervals contain the greatest frequencies.

---

## 3. Measures of Center: Mean vs. Median

### Mean
The mean is the average of all observations.

Formula:

Mean = Sum of all observations ÷ Number of observations

### Median
The median is the middle observation in an ordered dataset. If there is an even number of observations, the median is the average of the two middle values.

### Relationship Between Mean and Median

- **Right-skewed distribution:** Mean is typically greater than the median.
- **Left-skewed distribution:** Mean is typically less than the median.
- **Symmetric distribution:** Mean and median are typically approximately equal.

Note: These are general patterns, not strict rules.

---

## 4. Five-Number Summary

The five-number summary describes the center and spread of quantitative data.

1. **Minimum:** The smallest observation.
2. **First Quartile (Q1):** The 25th percentile; approximately 25% of observations are at or below this value.
3. **Median (Q2):** The 50th percentile; divides the ordered data into two halves.
4. **Third Quartile (Q3):** The 75th percentile; approximately 75% of observations are at or below this value.
5. **Maximum:** The largest observation.

### Interquartile Range (IQR)

The IQR measures the spread of the middle 50% of the data.

Formula:

IQR = Q3 − Q1

The IQR is resistant to extreme values and is useful for identifying potential outliers.

### Range

The range measures the difference between the largest and smallest observations.

Formula:

Range = Maximum − Minimum

Note: Range and IQR measure spread, while standard deviation measures spread around the mean.

---

## 5. Standard Deviation (SD)

Standard deviation measures how much observations typically vary around the mean.

- **Low SD:** Observations tend to cluster near the mean.
- **High SD:** Observations tend to be more spread out.

A standard deviation is not exactly the average distance from the mean, but it provides a useful measure of typical spread.

### Important
The mean and standard deviation are sensitive to outliers. The median and IQR are more resistant to outliers.

---

## 6. The Empirical Rule: 68–95–99.7

The Empirical Rule describes approximately how data are distributed around the mean in a **normal (bell-shaped) distribution**.

- **68%** of observations fall within 1 standard deviation of the mean.
- **95%** of observations fall within 2 standard deviations of the mean.
- **99.7%** of observations fall within 3 standard deviations of the mean.

In symbols:

- 68%: Mean ± 1 SD
- 95%: Mean ± 2 SD
- 99.7%: Mean ± 3 SD

Example:

If the mean is 100 and the standard deviation is 10:

- About 68% of observations fall between 90 and 110.
- About 95% fall between 80 and 120.
- About 99.7% fall between 70 and 130.

---

## 7. Z-Score (Standard Score)

### Definition
A Z-score measures how many standard deviations an observation is above or below the mean.

### Formula

For a population:

z = (x − μ) / σ

For a sample, when using the sample mean and sample standard deviation:

z = (x − x̄) / s

Where:

- **z:** Z-score
- **x:** Individual observation
- **μ:** Population mean
- **σ:** Population standard deviation
- **x̄:** Sample mean
- **s:** Sample standard deviation

### How to Interpret a Z-Score

- **Positive Z-score:** The observation is above the mean.
- **Negative Z-score:** The observation is below the mean.
- **Z-score = 0:** The observation equals the mean.
- **Large absolute Z-score:** The observation is farther from the mean.

Example:

If z = 2, the observation is 2 standard deviations above the mean.

If z = −1.5, the observation is 1.5 standard deviations below the mean.

---

## 8. Boxplots

### What Is a Boxplot?

A boxplot is a graphical representation of quantitative data based on the five-number summary.

It helps us understand:

- **Center:** The median.
- **Spread:** The interquartile range (IQR) and whisker lengths.
- **Outliers:** Unusually low or high observations.
- **Comparisons:** Differences in center and spread between two or more groups.

### Parts of a Boxplot

- **Left edge of the box:** Q1.
- **Line inside the box:** Median (Q2).
- **Right edge of the box:** Q3.
- **Box length:** IQR = Q3 − Q1.
- **Whiskers:** Extend to the most extreme observations that are not classified as outliers under the boxplot rule.
- **Points beyond the whiskers:** Potential outliers.

### The 1.5 × IQR Rule

Potential outliers are observations below or above these fences:

Lower fence = Q1 − 1.5 × IQR

Upper fence = Q3 + 1.5 × IQR

Any observation below the lower fence or above the upper fence is considered a potential outlier.

Note: In a standard modified boxplot, whiskers generally extend to the most extreme observations within these fences, not necessarily to the minimum and maximum.

### Comparing Boxplots

When comparing two or more groups, look at:

1. **Median:** Which group has the higher typical value?
2. **IQR:** Which group has more variability in its middle 50%?
3. **Overall spread:** Which group has greater variation?
4. **Outliers:** Does either group have unusual observations?
5. **Shape and skewness:** Is one group more asymmetric than another?

---

## 9. Quick Review

| Concept | What It Tells You |
|---|---|
| Mean | Average value |
| Median | Middle value |
| Range | Maximum − Minimum |
| IQR | Q3 − Q1; spread of the middle 50% |
| Standard deviation | Spread around the mean |
| Five-number summary | Minimum, Q1, median, Q3, maximum |
| Z-score | Number of standard deviations from the mean |
| Histogram | Distribution of quantitative data |
| Boxplot | Center, spread, and potential outliers |
| Bar chart | Comparison of categorical frequencies |
| Pie chart | Proportions of categorical data |
| Empirical Rule | 68%, 95%, and 99.7% within 1, 2, and 3 SDs of the mean for a normal distribution |

**Exam tip:** For symmetric data without extreme outliers, use the mean and standard deviation to describe center and spread. For skewed data or data with outliers, the median and IQR are usually more appropriate.
