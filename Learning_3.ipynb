# Linux Commands – Viewing and Analyzing Text Files

## 6. `ip a` — Display network interfaces and addresses


- `ip` — networking management tool.

## 7. `ip addr show eth0` — Inspect a specific interface

## 8. `ip -4 addr show` — Display IPv4 addresses

```bash
!ip -4 addr show
```

Shows network addresses using IPv4 only.

- `-4` — selects IPv4.
- IPv4 addresses look like `172.28.0.12`.

## 9. `ip route` — Display the routing table

```bash
!ip route
```

Shows how network traffic is routed.

Example output:
```text
default via 172.28.0.1 dev eth0
172.28.0.0/16 dev eth0 proto kernel scope link src 172.28.0.12
```

Meaning:
- `default via 172.28.0.1` — the default gateway used to reach destinations without a more specific route.
- `dev eth0` — traffic uses the `eth0` interface.
- `172.28.0.0/16` — the local network range.
- `src 172.28.0.12` — the preferred source IP address for that route.

# Python File Handling, TAR, and ZIP

## 1. Writing and Reading Files

**Write a file (`w`):** Creates a file or overwrites existing content.
```python
%%writefile q.txt
hi, i am milad ghavami, and i want to provide some details
```

**Read a file (`r`):** Reads the file's content.
```python
with open("q.txt", "r") as file:
    print(file.read())
```

**Append to a file (`a`):** Adds content to the end without deleting existing content.
```python
with open("q.txt", "a") as file:
    file.write("\nthis is the library")
```
- `\n` starts a new line.
- `with open(...)` automatically closes the file.
- `file.read()` reads the content; `file.write()` writes content.

**Display a file in Colab:**
```python
!cat q.txt
```

## 3. Creating a TAR Archive

```python
!tar -cvf /content/milad.tar /content/milad
```
- `tar`: Archive utility.
- `-c`: Create an archive.
- `-v`: Display files being processed.
- `-f`: Specify the archive filename.
- `.tar`: Groups files and folders together **without compression by default**.

**List archive contents without extracting:**
```python

```
- `-t`: List contents.
- `-v`: Show details.
- `-f`: Specify the archive.

- `-x`: Extract files.
- `-v`: Display extracted files.
- `-f`: Specify the archive.

**Extract into a specific directory:**

```python
!tar -xvf /content/milad.tar -C /content/my_destination
```

- `mkdir -p`: Creates the directory, including missing parent directories, without an error if it already exists.

- `zip`: Creates a ZIP archive.

# Bash Shell Scripts — Quick Notes

### 1. Create and run a shell script

A **Bash script** is a file containing commands that run in sequence.

```bash
!touch h.sh                         # Create an empty file
!echo '#!/bin/bash' >> h.sh          # Add the Bash interpreter line
!echo 'echo Hello World' >> h.sh    # Add a command to the file
!cat h.sh                            # Display the file contents
!chmod u+x h.sh                      # Give the owner execute permission
!./h.sh                              # Run the script
!which bash                          # Find the Bash executable
```

**Output of `cat h.sh`:**
```bash
#!/bin/bash
echo Hello World
```

**Output of `which bash`:** `/usr/bin/bash`

### 2. Important commands

- `touch filename.sh` — creates an empty file.
- `echo "text"` — prints text or writes it to a file.
- `>>` — appends text to a file without deleting existing content.
- `>` — overwrites a file's contents.
- `cat filename.sh` — displays file contents.
- `chmod u+x filename.sh` — gives the file owner permission to execute the file.
- `./filename.sh` — runs an executable script.
- `bash filename.sh` — runs a script using Bash without requiring execute permission.
- `which bash` — shows the location of the Bash executable.

**Note:** `!` runs a shell command from a Google Colab code cell. It is not part of normal Bash syntax.

### 3. Create an interactive greeting script

```bash
!touch greet.sh
```

Write the following code into `greet.sh`:

```bash
#!/bin/bash

# Ask the user for their name
echo -n "Enter your name :"
read name

# Display a personalized greeting
echo "Welcome $name"

# Print a message on one line
echo -n "Congratulations! You just created and ran your first shell script "
echo "using Bash on IBM Skills Network"
```

Make the script executable and run it:

```bash
!chmod u+x greet.sh
!bash greet.sh
```

**Example output:**
```text
Enter your name :milad
Welcome milad
Congratulations! You just created and ran your first shell script using Bash on IBM Skills Network
```

### 4. Key concepts

- `#!/bin/bash` — **shebang**; specifies Bash as the interpreter.
- `#` — starts a comment; comments explain the code and are not executed.
- `echo -n` — prints text without adding a newline.
- `read name` — accepts user input and stores it in the variable `name`.
- `$name` — accesses the value stored in the variable.
- `echo "Welcome $name"` — prints the greeting with the entered name.
- `chmod u+x` — `u` means owner, `+x` means add execute permission.
- `ls -l filename.sh` — displays file permissions and other file details. For example, `-rwxr--r--` indicates that the owner can read, write, and execute the file.

**Google Colab tip:** To create or overwrite a complete script more easily, use `%%writefile greet.sh` as the first line of a code cell, followed by the script contents. It writes the cell's contents to the file, replacing any existing contents.

### Summary

**Create → Write → Check → Make executable → Run**

Remember: `touch` creates, `echo` writes or prints, `cat` displays, `chmod` changes permissions, and `bash` executes the script.
