# Linux Informational Commands — Quick Notes

### 1. `whoami` — Current user
```bash
!whoami
```
Displays the current username. In Colab, the output is usually `root`.

### 2. `who` — Logged-in users
```bash
!who
```
Shows users currently logged into the system. It may return no output in Colab.

### 3. `uname` — Operating system kernel
```bash
!uname
```
Displays the kernel name, usually `Linux`.

### 4. `uname -a` — System information
```bash
!uname -a
```
Displays the kernel name, hostname, kernel version, architecture, and operating system details.

### 5. `date` — Current date and time
```bash
!date
```
Displays the current date and time. Colab usually uses UTC by default.

### 6. Display Pacific Time
```bash
!TZ="America/Los_Angeles" date "+%m/%d/%Y %I:%M:%S %p %Z"
```
Displays the date and time in Pacific Time, automatically accounting for daylight saving time.

**Format symbols:**
- `%m` = month
- `%d` = day
- `%Y` = four-digit year
- `%I` = hour (12-hour format)
- `%M` = minutes
- `%S` = seconds
- `%p` = AM/PM
- `%Z` = timezone abbreviation (PDT or PST)
- `+` = tells `date` to use the specified output format
- `/` and `:` = separators between values

**Example output:** `10/11/2026 10:59:07 AM PDT`

**Remember:** In Colab, prefix shell commands with `!`. The `TZ="America/Los_Angeles"` setting affects only that command, not the entire Colab environment.

# Linux File and Directory Permissions — Quick Notes

## 1. Create a directory
```bash
!mkdir -p data
```
Creates `data`. `-p` also creates missing parent directories without an error if the directory already exists.

## 2. List files and change directories
```bash
!ls
%cd data
!pwd
```
- `ls`: lists files and folders.
- `%cd data`: changes the current directory in Colab.
- `pwd`: shows the current directory (`/content/data`).

## 3. Download a file
```bash
!wget -O usdoi.txt "URL"
```
Downloads a file and saves it as `usdoi.txt`. `-O` specifies the output filename.

## 4. View file permissions
```bash
!ls -l usdoi.txt
```
Example: `-rw-r--r--`
- First character: `-` = regular file; `d` = directory.
- `rw-` = owner can read and write.
- `r--` = group can read.
- `r--` = others can read.
- `-` means the permission is not granted.

## 5. Change permissions with `chmod`
```bash
!chmod g+w usdoi.txt
!chmod o+x usdoi.txt
!chmod u-r usdoi.txt
```
- `chmod`: changes permissions.
- `u` = owner, `g` = group, `o` = others, `a` = all.
- `+` adds a permission; `-` removes one.
- `r` = read, `w` = write, `x` = execute.

Use `!ls -l usdoi.txt` after each command to verify the result.

## 6. Directory permissions
```bash
%cd /content
!mkdir -p test
!chmod u-x data
!ls -ld data
```
- `ls -ld data`: displays details about the directory itself.
- For directories, `r` allows listing contents, `w` allows changing entries, and `x` allows entering/traversing the directory.
- Removing `x` can prevent normal access, even if `r` or `w` remains.

## 7. Important reminders
- `!` runs a shell command in Colab.
- `%cd` changes the notebook's working directory.
- `-l` means detailed listing.
- `-d` means display the directory itself.
- Colab often runs commands as `root`, so permission-denied examples may behave differently than in a regular Linux terminal.

# Linux Commands – Viewing and Analyzing Text Files

## 1. `more` – View a file page by page
```bash
!more usdoi.txt
```
Displays the file one screen at a time. Press **Space** to see the next page and **q** to quit.

## 2. `less` – Scroll through a file
```bash
!less usdoi.txt
```
Allows you to move up and down through the file. Use the arrow keys or Page Up/Page Down, and press **q** to quit.

## 3. `head` – Display the beginning of a file
```bash
!head usdoi.txt
```
Displays the first **10 lines** by default.

```bash
!head -3 usdoi.txt
```
Displays only the first **3 lines**. The number after `-` specifies how many lines to show.

## 4. `tail` – Display the end of a file
```bash
!tail -2 usdoi.txt
```
Displays the last **2 lines** of the file.

## 5. `wc` – Count file contents
```bash
!wc -l usdoi.txt
```
Counts lines. The result shown is **152 lines**.

```bash
!wc -w usdoi.txt
```
Counts words. The result shown is **1,330 words**.

Other useful options:
- `wc -c usdoi.txt` — counts bytes.
- `wc -m usdoi.txt` — counts characters.

## 6. `sort` – Sort lines alphabetically
```bash
!sort usdoi.txt
```
Displays the file's lines in sorted order. It does not modify the original file.

## 7. `sort -r` – Reverse the sorting order
```bash
!sort -r usdoi.txt
```
Displays the lines in reverse sort order.

## 8. `grep` – Search for matching text
```bash
!grep people usdoi.txt
```
Displays every line containing the text `people`.

Useful options:
- `grep -i people usdoi.txt` — ignores uppercase/lowercase differences.
- `grep -n people usdoi.txt` — displays matching line numbers.
- `grep -v people usdoi.txt` — displays lines that do not contain `people`.
- `grep -c people usdoi.txt` — counts matching lines.

# Linux Networking Commands — Study Notes

## 1. `hostname` — Show the computer's name

```bash
!hostname
```

Displays the hostname of the current computer or container.

Example output: `da57ac8c0215`

## 2. `hostname -i` — Show the hostname's IP address

```bash
!hostname -i
```

Displays the IP address associated with the hostname.

Example: `172.28.0.12`

## 3. `%bash` — Run multiple Bash commands

```bash
%%bash
echo "Hello from Linux"
pwd
whoami
```

In Google Colab, `%%bash` must be the **first line of the cell**. It runs the entire cell as a Bash script.

- `echo` — prints a message.
- `pwd` — displays the current working directory.
- `whoami` — displays the current username.

**Note:** `%bash` and `%%bash` are different. In Colab, use `%%bash` for a Bash cell.

## 4. `hostname -I` — Show all assigned IP addresses

```bash
!hostname -I
```

Displays the IP addresses assigned to the computer, separated by spaces. It may show more than one address.

Difference:
- `hostname -i` — resolves the hostname to an IP address.
- `hostname -I` — displays assigned IP addresses.

## 5. `sudo apt install iproute2` — Install networking tools

```bash
!sudo apt install iproute2
```

- `sudo` — runs a command with administrator privileges.
- `apt` — manages software packages on Debian/Ubuntu Linux.
- `install` — installs a package.
- `iproute2` — provides networking tools such as `ip`.

## 6. `ip a` — Display network interfaces and addresses

```bash
!ip a
```

Shows network interfaces, their status, and assigned IP addresses.

- `ip` — networking management tool.
- `a` — abbreviation for `address`.

## 7. `ip addr show eth0` — Inspect a specific interface

```bash
!ip addr show eth0
```

Displays network information for the interface named `eth0`, including its IP address and status.

## 8. `ip -4 addr show` — Display IPv4 addresses

```bash
!ip -4 addr show
```

Shows network addresses using IPv4 only.

- `-4` — selects IPv4.
- IPv4 addresses look like `172.28.0.12`.

## 9. `ip route` — Display the routing table

```bash
!ip route
```

Shows how network traffic is routed.

Example output:
```text
default via 172.28.0.1 dev eth0
172.28.0.0/16 dev eth0 proto kernel scope link src 172.28.0.12
```

Meaning:
- `default via 172.28.0.1` — the default gateway used to reach destinations without a more specific route.
- `dev eth0` — traffic uses the `eth0` interface.
- `172.28.0.0/16` — the local network range.
- `src 172.28.0.12` — the preferred source IP address for that route.

## 10. `ip --version` — Why it gives an error

```bash
!ip --version
```

The `ip` command may not support `--version`, so it reports an unknown option.

To check available options, use:

```bash
!ip -help
```

## 11. `curl` — Retrieve data from a URL

```bash
!curl "https://example.com"
```

`curl` transfers data from or to a URL. Without extra options, it usually prints the response body in the output.

Your screenshot uses a course URL ending in `usdoi.txt` to retrieve the file's contents.

To save a downloaded response to a file, use `-o`:

```bash
!curl -o usdoi.txt "https://example.com/usdoi.txt"
```

Replace the example URL with the actual URL you need.

## 12. `sudo apt update` — Update package lists

```bash
!sudo apt update
```

Refreshes the local list of available software packages. It does not upgrade all installed packages.

## 13. `sudo apt install -y iputils-ping` — Install ping

```bash
!sudo apt install -y iputils-ping
```

Installs the package that provides the `ping` command.

- `-y` — automatically answers yes to confirmation prompts.

## 14. `ping -c 5 www.google.com` — Test network connectivity

```bash
!ping -c 5 www.google.com
```

Sends five ICMP echo requests to Google's hostname to test whether it is reachable and measure response times.

- `ping` — tests network reachability.
- `-c 5` — sends five requests, then stops.
- `www.google.com` — the destination hostname.

If it works, you should see reply messages and packet statistics. Failure does not always mean the internet is unavailable; ICMP traffic or DNS may be blocked.
